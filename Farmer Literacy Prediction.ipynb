```python
import numpy as np
import pandas as pd

np.random.seed(43)  # For reproducibility

n_samples = 500  # Simulated farmers

# Simulated features....Generating Random Dataset
age = np.random.randint(18, 70, n_samples) #Generating 500 random ages
literacy_level = np.random.choice([1, 2, 3], n_samples, p=[0.5, 0.35, 0.15])  # Assisning literacy level, 1=Low, 2=Medium, 3=High
farming_experience_years = np.round(np.random.uniform(1, 25, n_samples), 1) #Generating and Assigning experience yeaar to each farmers rounding up to whole number
tech_exposure_level = np.random.choice([0, 1, 2], n_samples, p=[0.6, 0.3, 0.1])  #Assigning the familiarity to tech level, 0=No experience, 1=Basic, 2=Advanced
native_language_support = np.random.choice([0, 1], n_samples, p=[0.3, 0.7])  # 1 means farmers need AI supports or native language
avg_task_completion_time = np.round(np.random.uniform(10, 60, n_samples), 2)  # seconds used by a farmeer to complete a tast, interacting with  computer
error_rate = np.round(np.random.uniform(0, 0.8, n_samples), 2)#Assigning the rate at which errors is being made

# Target: Will user struggle and need more AI voice assistance?
need_voice_assist = (
    (literacy_level == 1) & (tech_exposure_level == 0) & (error_rate > 0.4) & (avg_task_completion_time > 40)
).astype(int)#If all this conditions are met that means the farmer needs a voic eassistant

# Create dataset
df = pd.DataFrame({
    "Age": age,
    "Literacy_Level": literacy_level,
    "Farming_Experience_Years": farming_experience_years,
    "Tech_Exposure_Level": tech_exposure_level,
    "Native_Language_Support": native_language_support,
    "Task_Completion_Time": avg_task_completion_time,
    "Error_Rate": error_rate,
    "Need_Voice_Assistance": need_voice_assist
})

print(df.head())
print("\nDataset Shape:", df.shape)
print("Number needing voice assistance:", df['Need_Voice_Assistance'].sum())
/
```

```python
#IMPORTING NECCESSARY LIBRARIES
import pandas as pd #for data cleaning
import numpy as np #for numerical data
from sklearn.model_selection import train_test_split #for trainiing and testing dataset
from sklearn.preprocessing import StandardScaler #for transforming the dataset
from sklearn.linear_model import LogisticRegression #for creating model
from sklearn.metrics import accuracy_score, confusion_matrix,classification_report #Generating insight about the model


x = df.drop("Need_Voice_Assistance", axis = 1) #The independent dataset
y = df["Need_Voice_Assistance"] #The targetted dataset
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size = 0.20, random_state = 42) #splittiing the dataset into training  and testing dataset

#for transforming our independennt dataset
scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)

#Creating our Prediction Model    
model = LogisticRegression()
model.fit(x_train, y_train) #Training the model
y_pred = model.predict(x_test)#Testing the model

print("Model Accuracy : ", accuracy_score(y_test, y_pred)) #Evaluating how accurate the model is
print("\n🔍 Classification Report:\n", classification_report(y_test, y_pred)) #report.....
print("\n🧱 Confusion Matrix:\n", confusion_matrix(y_test, y_pred))#full analysis of the model prediction
sample_farmer = np.array([[34, 1, 3, 0, 1, 55, 0.6]])  # Simulated farmer record
sample_farmer = scaler.transform(sample_farmer)  #Transforming the simulated farmer record

prediction = model.predict(sample_farmer)  # Predicting the need for assistance for the simulated farmer record

print("\n🎙 Interface Adaptation Decision:") #Display statement if prediction is True
if prediction[0] == 1:
    print("⚠ Farmer likely to struggle → Use VOICE-FIRST interface")
    print("- Slow speech speed")
    print("- Use native language")
    print("- Give confirmation prompts")
else:  #Display statement if prediction is False
    print("✔ Farmer likely comfortable → Use standard interface")
```

```python
# Simulated decision logic from Phase 3

# Prediction from our ML model (Simulated)
prediction = 1  # 1 means user needs assistance, 0 means regular UI

def adaptive_interface(prediction): #Defining adaptive interface function
    if prediction == 1:  #Return statement if prediction is True
        return {
            "Interface_Type": "Voice-First",
            "Accessibility_Mode": "High",
            "Instruction_Delivery": "Step-by-Step Verbal Guidance",
            "Language": "Localized (e.g., Yoruba/Hausa/Igbo)",
            "Extra_Features": ["Error correction hints", "Slow interaction speed"],
        }
    else:  #Rreturn statement if prediction is otherwise
        return {
            "Interface_Type": "Standard UI",
            "Accessibility_Mode": "Normal",
            "Instruction_Delivery": "On-screen prompts",
            "Language": "English (default)",
            "Extra_Features": ["Quick navigation", "Advanced options"],
        }

# Simulate decision
interface_config = adaptive_interface(prediction)
print("Adaptive Interface Config:", interface_config)
```
